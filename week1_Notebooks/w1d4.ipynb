# W1D4: Exploratory Data Analysis (EDA)

This notebook analyses a reproducible 20-student dataset. It runs `df.describe()`, `df.info()`, and `df.isnull().sum()`, imputes missing ML scores using the median, engineers average-score grades, and writes the required plots to `artifacts/eda`.

```python
from pathlib import Path
import sys

repository_root = Path.cwd().resolve()
if not (repository_root / 'src').exists():
    repository_root = repository_root.parent
if str(repository_root) not in sys.path:
    sys.path.insert(0, str(repository_root))

from src.w1d4_eda import (
    create_student_dataset, engineer_features, inspect_data,
    plot_correlation_heatmap, plot_numeric_distributions,
    plot_top_categories, print_feature_summary,
)
```

```python
raw_df = create_student_dataset()
inspect_data(raw_df)

print('=== Dataset Overview ===')
display(raw_df.head())
print(f'Shape: {raw_df.shape}')
print('\ndf.describe()')
display(raw_df.describe(include='all'))
print('\ndf.info()')
raw_df.info()
print('\ndf.isnull().sum()')
display(raw_df.isnull().sum())
```

```python
df = engineer_features(raw_df)

print('=== Grade Distribution ===')
display(df['grade'].value_counts().sort_index())
print('=== Top 5 Students ===')
display(df.nlargest(5, 'avg_score')[['name', 'avg_score', 'grade']])
print('=== Attendees vs Non-Attendees ===')
display(df.groupby('attended')['avg_score'].agg(['mean', 'count']).round(2))
```

```python
plot_numeric_distributions(df)
plot_correlation_heatmap(df)
plot_top_categories(df)

print('Saved: artifacts/eda/numeric_distributions.png')
print('Saved: artifacts/eda/correlation_heatmap.png')
print('Saved: artifacts/eda/top_10_category_counts.png')
```

## Five observations

1. There are 20 students and three missing values, all in `ml_score`.
2. Ages range only from 21 to 25.
3. Median imputation keeps all 20 records for the score analysis.
4. Grades are concentrated in C (11 students) and B (7 students); no student receives A.
5. Attendance groups are imbalanced (16 attendees and 4 non-attendees), so their average-score difference is not a reliable causal finding.
