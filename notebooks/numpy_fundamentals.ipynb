# NumPy Fundamentals

This notebook demonstrates arrays, broadcasting, vectorization, and CSV-based statistics.

```python
import numpy as np

vector = np.array([10, 20, 30, 40])
matrix = np.array([[1, 2, 3], [4, 5, 6]])
tensor = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print('1D shape:', vector.shape)
print('2D shape:', matrix.shape)
print('3D shape:', tensor.shape)
```

```python
base = np.array([10, 20, 30], dtype=np.float64)
scaled = base * 1.5
print('Broadcasted result:', scaled)
```

```python
marks = np.array([55, 67, 89, 91, 72])
weights = np.array([0.8, 0.9, 1.0, 1.1, 1.2])
weighted = marks * weights
print('Weighted scores:', weighted)
print('Mean:', np.mean(weighted))
print('Std:', np.std(weighted))
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print('A @ B =\n', A @ B)
```

```python
data = np.genfromtxt('data/marketing_metrics.csv', delimiter=',', skip_header=1, names=True)
spend = data['ad_spend']
leads = data['leads']
roi = data['roi']
print('Mean ad spend:', np.mean(spend))
print('Std ad spend:', np.std(spend))
print('Correlation spend vs leads:', np.corrcoef(spend, leads)[0, 1])
print('Correlation leads vs roi:', np.corrcoef(leads, roi)[0, 1])
```
